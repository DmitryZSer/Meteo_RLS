```python
def compute_fbar_ns(v, w, dxy, V_a, g, L):
    delta_v = np.zeros_like(v)
    half_L = L / 2  # 0.5 km
    
    num_points = int(half_L / dy)
    
    padded_v = np.pad(v, ((0, 0), (num_points, num_points), (0, 0)), mode='constant')
    
    delta_v = padded_v[:, num_points:-num_points, :] - padded_v[:, :-2*num_points, :]
    
    term1 = (V_a / (g * L)) * delta_v

    sum_w = np.sum(w[:, :num_points*2, :], axis=1)  
    term2 = (dxy / (L * V_a)) * sum_w
    term2 = np.expand_dims(term2, axis=1) 
    term2 = np.repeat(term2, term1.shape[1], axis=1)
    
    fbar_ns = term1 - term2
    
    return fbar_ns

fbar_ns = compute_fbar_ns(v_data['data'], w_data['data'], dxy, V_a, g, L)
```

```python
def compute_fbar_ew(u, w, dxy, V_a, g, L):
    delta_u = np.zeros_like(u)
    half_L = L / 2  # 0.5 km
    
    num_points = int(half_L / dx)
    
    padded_u = np.pad(u, ((num_points, num_points), (0, 0), (0, 0)), mode='constant')
    
    delta_u = padded_u[num_points:-num_points, :, :] - padded_u[:-2*num_points, :, :]
    
    term1 = (V_a / (g * L)) * delta_u

    sum_w = np.sum(w[:num_points*2, :, :], axis=0)  
    term2 = (dxy / (L * V_a)) * sum_w
    

    fbar_ew = term1 - term2
    
    return fbar_ew

fbar_ew = compute_fbar_ew(u_data['data'], w_data['data'], dxy, V_a, g, L)
```

```python
z_slice = 6
y_slice = 80
plt.figure(figsize=(10, 7))
plt.imshow(fbar_ns[:, :, z_slice], cmap='coolwarm', origin='lower')
plt.colorbar(label='FBAR_N-S (m/s)')
plt.title(f'FBAR N-S (z={z_slice})')
plt.xlabel('X Index')
plt.ylabel('Y Index')
plt.show()

plt.figure(figsize=(10, 7))
plt.imshow(fbar_ew[:, :, z_slice], cmap='coolwarm', origin='lower')
plt.colorbar(label='FBAR_E-W (m/s)')
plt.title(f'FBAR  E-W (z={z_slice})')
plt.xlabel('X Index')
plt.ylabel('Y Index')
plt.show()
```

```python
# u_data = read_pws_file("../data/111/PWS_NASA_Dataset111_u.faa")
# v_data = read_pws_file("../data/111/PWS_NASA_Dataset111_v.faa")
# w_data = read_pws_file("../data/111/PWS_NASA_Dataset111_w.faa")

u_data = read_pws_file("../data/DO-220B Reference Supplement/ADWRS/PWS_NASA_Dataset237_W9_2/PWS_NASA_Dataset237_u.faa")
v_data = read_pws_file("../data/DO-220B Reference Supplement/ADWRS/PWS_NASA_Dataset237_W9_2/PWS_NASA_Dataset237_v.faa")
w_data = read_pws_file("../data/DO-220B Reference Supplement/ADWRS/PWS_NASA_Dataset237_W9_2/PWS_NASA_Dataset237_w.faa")

U = u_data['data']  # Horizontal wind component (m/s)
W = w_data['data']  # Vertical wind component (m/s)
dx = u_data['resolution_m'][0] 
g = 9.81  # Gravitational acceleration (m/s²)
v = 77.2  # Aircraft true airspeed (m/s, fixed at 150 knots)
[]
dU_dx = np.zeros_like(U)
dU_dx[:-1, :, :] = (U[1:, :, :] - U[:-1, :, :]) / dx
dU_dx[-1, :, :] = 0  # Assume zero gradient at last point

```

```python
# (dU/dx * v) / g
term1 = (dU_dx * v) / g

# W / v
term2 = W / v

F = term1 - term2

FBAR = np.zeros_like(F)
half_L = 6 # grid points on each side (total n points = 1 km)

for i in range(U.shape[0]):
    start = max(0, i - half_L)
    end = min(U.shape[0], i + half_L + 1)
    FBAR[i, :, :] = np.mean(F[start:end, :, :], axis=0)

z_slice = 6
FBAR[:, :, z_slice].max()
```

```python

FBAR_slice = FBAR[:, :, z_slice]  

x = origin[0] + np.arange(ix) * dx  
y = origin[1] + np.arange(iy) * dy  
X, Y = np.meshgrid(x, y, indexing='ij')

fig = go.Figure()

fig.add_trace(go.Heatmap(
    x=x,
    y=y,
    z=FBAR_slice,
    colorscale='Viridis',
    zmin=np.nanpercentile(FBAR_slice, 5),  
    zmax=np.nanpercentile(FBAR_slice, 95),
    hovertemplate='X: %{x:.1f} м<br>Y: %{y:.1f} м<br>FBAR: %{z:.2f}',
    colorbar=dict(title="F-factor (1/с)")
))

fig.update_layout(
    title=f"Распределение F-factor (FBAR) на высоте Z={z[z_slice]:.1f} м",
    xaxis_title="Координата X (м)",
    yaxis_title="Координата Y (м)",
    template="plotly_white",
    width=800,
    height=800,
    scene_aspectmode='data'
)

fig.show()
```
